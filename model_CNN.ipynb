```python
from fastai.vision.all import *

print("กำลังโหลดรูปภาพและเตรียมสมอง AI...")

path = Path('House_Room_Dataset')


dls = ImageDataLoaders.from_folder(
    path, 
    valid_pct=0.2,       
    seed=42, 
    item_tfms=Resize(224), 
    bs=32                 
)


print("\nตัวอย่างรูปภาพที่ AI มองเห็น:")
dls.show_batch(max_n=9)


learn = vision_learner(dls, resnet18, metrics=accuracy)


print("\n กำลังเริ่มเทรน AI") 
learn.fine_tune(3) 

print("\n เทรนเสร็จสมบูรณ์")
```

```python
learn.export('room_classifier_fastai.pkl')

print(" เซฟไฟล์สมอง AI (FastAI) ")
```
